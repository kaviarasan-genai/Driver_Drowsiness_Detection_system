```python
# Importing Libraries
import numpy as np
import pandas as pd
import os
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.layers import Dense, Dropout, Activation, Flatten, Conv2D, MaxPooling2D
```

```python
# Split the dataset into Two files for Training-80% & Testing-20%
import splitfolders
splitfolders.ratio(r"C:\Users\Asus\OneDrive\Desktop\Driver_Detection\driver_datas",output = "split_data",seed = 42,ratio = (0.8, 0.2))
```

```python
# Train & Test split file to Generate Images by applying a Model for Understanding Patterns
training_data = ImageDataGenerator(rescale = 1.0/255)
train_data = training_data.flow_from_directory(r"C:\Users\Asus\OneDrive\Desktop\Driver_Detection\split_data\train",
                                                target_size = (224, 224),
                                                batch_size = 8,
                                                class_mode = "categorical")


test_data = training_data.flow_from_directory(r"C:\Users\Asus\OneDrive\Desktop\Driver_Detection\split_data\val",
                                             target_size = (224, 224),
                                             batch_size = 8,
                                             class_mode = "categorical")
```

```python
# Train the CNN Model by using Relu & Softmax Activation Function

model = tf.keras.models.Sequential()

model.add(tf.keras.layers.Conv2D(filters=32, padding = "same", kernel_size = 3, activation = "relu", input_shape = [224, 224, 3]))
model.add(tf.keras.layers.MaxPool2D(pool_size = 2, strides = 2))

model.add(tf.keras.layers.Conv2D(filters = 64, padding = "same", kernel_size = 3, activation = "relu"))
model.add(tf.keras.layers.MaxPool2D(pool_size = 2, strides = 2))

model.add(tf.keras.layers.Flatten())


model.add(tf.keras.layers.Dense(units = 64, activation = "relu"))
model.add(tf.keras.layers.Dense(units = 32, activation = "relu"))
model.add(tf.keras.layers.Dropout(0.5))
model.add(tf.keras.layers.Dense(units = 3, activation = "softmax"))
```

```python
model.summary()
```

```python
# Compile a model

model.compile(optimizer= tf.keras.optimizers.Adam(learning_rate=0.001), loss = "categorical_crossentropy", metrics = ["accuracy"])
```

```python
# Fit the Model to get Accuracy and Loss

models = model.fit(train_data, validation_data = test_data, epochs = 5)
print(models)
```

```python
# Model Final Accuracy and Loss

loss , accuracy = model.evaluate(test_data)

print("="*40)
print("  Model Accuracy  :", accuracy)
print("  Model Loss      :", loss)
print("="*40)
```

```python
# Give New Image Input to Check the Model Prediction Output

class_name = train_data.class_indices

img       = tf.keras.utils.load_img(r"C:\Users\Asus\OneDrive\Desktop\Driver_Detection\driver_datas\eyes_closed\frame_09.jpg", target_size = (224, 224))
img_array = tf.keras.utils.img_to_array(img)
img_pix   = img_array/255.0
img_dim   = np.expand_dims(img_pix, axis = 0)

pred      = model.predict(img_dim)

print(pred)
print(class_name)
```
