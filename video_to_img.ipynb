```python
# In a video each second contains one frames to save
import os
import cv2

video_path = r"C:\Users\Asus\OneDrive\Desktop\driver_ipyn\testing_video\9519998-uhd_4096_2160_25fps.mp4"
output_folder = r"video_frames_1"  

os.makedirs(output_folder, exist_ok=True)

cap = cv2.VideoCapture(video_path)
fps = cap.get(cv2.CAP_PROP_FPS)
frame_id = 0
saved = 0

while cap.isOpened():
  ret, frame = cap.read()
  if not ret:
    break
  if int(fps) > 0 and frame_id % int(fps) == 0:
    file_path = os.path.join(output_folder, f"frame_{frame_id}.jpg")
    cv2.imwrite(file_path, frame)
    saved += 1
  frame_id += 2
cap.release()

print(f"successfully saved {saved} frames to '{output_folder}' folder!")
```
